```python

from openai import OpenAI
import os

client = OpenAI(
    api_key=os.environ.get("GROQ_API_KEY"),
    base_url="https://api.groq.com/openai/v1",
)

response = client.responses.create(
    input="Explain the importance of fast language models",
    model="openai/gpt-oss-20b",
)
print(response.output_text)
```

```python
chat_completion = client.chat.completions.create(
    messages=[
        {"role": "user", "content": "What is the weather today in Kolkata, India?"}
    ],
    model="openai/gpt-oss-20b",
)

print(chat_completion.choices[0].message.content)
```

```python
from langchain_groq  import ChatGroq

llm = ChatGroq(model="openai/gpt-oss-20b", temperature=0.7, api_key=os.environ.get("GROQ_API_KEY"))
llm.invoke("What is the capital for West Bengal?").content
```
